```python
aspirin = {"name": "Aspirin", "mw": 180.16, "logp": 1.2, "hbd": 1, "hba": 4}
print(aspirin["mw"])
aspirin["hba"] = 3        # تغییر یک مقدار
print(aspirin)
```

```python
# Check whether aspirin's molecular weight passes Lipinski's MW limit (<= 500 Da)
if aspirin["mw"] <= 500:
    print("Molecular weight is acceptable")
else:
    print("Molecular weight is too high")
```

```python
# Define a function that counts how many Lipinski rule-of-five criteria a molecule violates
def lipinski_violations(mol):
    violations = 0
    if mol["mw"] > 500:
        violations += 1
    if mol["logp"] > 5:
        violations += 1
    if mol["hbd"] > 5:
        violations += 1
    if mol["hba"] > 10:
        violations += 1
    return violations

print(lipinski_violations(aspirin))
```

```python
# Apply the Lipinski filter to a list of drugs and report PASS/FAIL for each (values are approximate)
molecules = [
    {"name": "Aspirin",      "mw": 180.16, "logp": 1.2,  "hbd": 1, "hba": 4},
    {"name": "Ibuprofen",    "mw": 206.28, "logp": 3.97, "hbd": 1, "hba": 2},
    {"name": "Cyclosporine", "mw": 1202.6, "logp": 2.9,  "hbd": 5, "hba": 12},
    {"name": "Paracetamol",    "mw": 151.16, "logp": 0.5, "hbd": 2,  "hba": 2},
    {"name": "Amphotericin B", "mw": 924.1,  "logp": 0.8, "hbd": 12, "hba": 18},
]

for mol in molecules:
    v = lipinski_violations(mol)
    status = "PASS" if v <= 1 else "FAIL"
    print(f"{mol['name']}: {v} violations -> {status}")
```
